# Catagorical Analysis

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print("\n", col)
    print(df[col].value_counts())
```

```python
pd.crosstab(df['gender'], df['promoted'], normalize='index') * 100
```

```python
gender_promotion = (
    df.groupby('gender')['promoted']
    .mean() * 100
)

gender_promotion.plot(
    kind='bar',
    figsize=(7,5)
)

plt.title("Promotion Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Promotion Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

# Purpose

# Male/Female promotion rate compare panna.

# Promotion Rate by Department

```python
pd.crosstab(df['department'], df['promoted'], normalize='index') * 100
```

```python
department_promotion = (
    df.groupby('department')['promoted']
    .mean()
    .sort_values(ascending=False) * 100
)

department_promotion.plot(
    kind='bar',
    figsize=(9,5)
)

plt.title("Promotion Rate by Department")
plt.xlabel("Department")
plt.ylabel("Promotion Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

# Purpose

# Which department has higher/lower promotion rate nu identify panna.

# Promotion Rate by Education Level

```python
pd.crosstab(df['education_level'], df['promoted'], normalize='index') * 100
```

```python
education_promotion = (
    df.groupby('education_level')['promoted']
    .mean()
    .sort_values(ascending=False) * 100
)

education_promotion.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Promotion Rate by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Promotion Rate (%)")
plt.xticks(rotation=45)
plt.show()
```
